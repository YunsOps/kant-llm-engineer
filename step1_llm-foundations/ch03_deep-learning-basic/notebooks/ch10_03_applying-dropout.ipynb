```python
import torch
import torch.nn as nn

torch.manual_seed(0)

dropout = nn.Dropout(p=0.5)

x = torch.ones(10)

dropout.train()
y_train = dropout(x)

dropout.eval()
y_eval = dropout(x)

print(f"input: {x}")
print(f"train output: {y_train}")
print(f"eval output: {y_eval}")
```

```python
import torch
import torch.nn as nn

class DropoutMLP(nn.Module):
    def __init__(self, input_dim=784, hidden_dim=128, num_classes=10, dropout_p=0.3):
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.dropout = nn.Dropout(p=dropout_p)
        self.fc2 = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        x = self.flatten(x)
        x = self.fc1(x)
        x = self.relu(x)
        x = self.dropout(x)
        logits = self.fc2(x)
        return logits
```

```python
model = DropoutMLP(dropout_p=0.5)
images = torch.randn(4, 1, 28, 28)

model.train()
out_train_1 = model(images)
out_train_2 = model(images)

model.eval()
out_eval_1 = model(images)
out_eval_2 = model(images)

print(f"train outputs same?: {torch.allclose(out_train_1, out_train_2)}")
print(f"eval outputs same?: {torch.allclose(out_eval_1, out_eval_2)}")
```
