```python
import torch
from torch.utils.data import TensorDataset, DataLoader

X = torch.randn(64, 10)
y = torch.randint(0, 3, size=(64,), dtype=torch.long)

dataset = TensorDataset(X, y)
loader = DataLoader(dataset, batch_size=16, shuffle=True)
```

```python
def inspect_first_batch(loader):
    batch = next(iter(loader))
    print(f"batch type: {type(batch)}")

    if isinstance(batch, (tuple, list)):
        for i, item in enumerate(batch):
            print(f"batch[{i}] type: {type(item)}")
            if torch.is_tensor(item):
                print(f"batch[{i}] shape: {item.shape}")
                print(f"batch[{i}] dtype: {item.dtype}")
                print(f"batch[{i}] device: {item.device}")
                print()
    else:
        print(batch)

inspect_first_batch(loader)

batch_x, batch_y = next(iter(loader))
assert torch.isfinite(batch_x).all(), "입력에 Nan/Inf가 있습니다."
assert torch.isfinite(batch_y.float()).all(), "label에 Nan/Inf가 있습니다."
print(f"x min/max: {batch_x.min().item()}/{batch_x.max().item()}")
print(f"label counts: {torch.bincount(batch_y)}")
```

```python
import torch.nn as nn

model = nn.Linear(10, 3)
loss_fn = nn.CrossEntropyLoss()

batch_x, batch_y = next(iter(loader))

logits = model(batch_x)
loss = loss_fn(logits, batch_y)

print(f"batch_x shape: {batch_x.shape}") # 16, 10
print(f"batch_y shape: {batch_y.shape}") # 16, 
print(f"logits shape: {logits.shape}")   # 16, 3
print(f"loss: {loss.item()}")
```

```python
import torch
import torch.nn
from torch.utils.data import TensorDataset, DataLoader

# 아래 조건을 만족하는 데이터 파이프라인 점검 코드를 작성하세요.
# 1. 샘플 50개, feature 12개, class 4개인 Dataset을 만듭니다.
# 2. batch size 10인 DataLoader를 만듭니다.
# 3. 첫 batch의 shape, dtype, device를 출력합니다.
# 4. `nn.Linear(12, 4)` 모델에 batch를 넣어 logits shape를 확인합니다.
# 5. `CrossEntropyLoss`로 loss를 계산합니다.

X = torch.randn(50, 12)
y = torch.randint(0, 4, size=(50,), dtype=torch.long)
dataset = TensorDataset(X, y)
loader = DataLoader(dataset, batch_size=10, shuffle=True)

batch_x, batch_y = next(iter(loader))
print(f"batch_x - shape: {batch_x.shape}, dypte: {batch_x.dtype}, device: {batch_x.device}")
print(f"batch_y - shape: {batch_y.shape}, dypte: {batch_y.dtype}, device: {batch_y.device}")

model = nn.Linear(12, 4)
logits = model(batch_x)
print(f"logits shape: {logits.shape}")

loss_fn = nn.CrossEntropyLoss()
loss = loss_fn(logits, batch_y)
print(f"loss: {loss}")
```
