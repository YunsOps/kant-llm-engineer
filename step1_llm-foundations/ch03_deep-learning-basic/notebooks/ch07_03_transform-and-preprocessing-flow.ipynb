```python
import torch
from torch.utils.data import Dataset

class DummyImageDataset(Dataset):
    def __init__(self, images, labels, transform=None):
        assert len(images) == len(labels)
        self.images = images
        self.labels = labels

        self.transform = transform
    def __len__(self):
        return len(self.images)
    
    def __getitem__(self, idx):
        image = self.images[idx]
        label = self.labels[idx]

        if self.transform is not None:
            image = self.transform(image)

        return image, label
```

```python
class SimpleNormalize:
    def __call__(self, x):
        return (x -x.mean() / (x.std() + 1e-6))

images = torch.randn(10, 3, 32, 32)
labels = torch.randint(0, 2, size=(10,), dtype=torch.long)

dataset = DummyImageDataset(images, labels, transform=SimpleNormalize())

image, label = dataset[0]
print(f"image shape: {image.shape}")
print(f"image mean: {image.mean().item()}")
print(f"image std: {image.std().item()}")
print(f"label: {label}")
```
