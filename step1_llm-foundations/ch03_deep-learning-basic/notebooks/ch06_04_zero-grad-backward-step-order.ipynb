```python
import torch
import torch.nn as nn
import torch.optim as optim

x = torch.randn(4, 2)
y = torch.randn(4, 1)

model = nn.Linear(2, 1)
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

optimizer.zero_grad()
pred = model(x)
loss = loss_fn(pred, y)
loss.backward()
optimizer.step()

print(f"loss: {loss.item()}")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(42)

x = torch.randn(4, 2)
y = torch.randn(4, 1)

model = nn.Linear(2, 1)
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# print(model.weight)
weight_before = model.weight.detach().clone()
# print(weight_before)

optimizer.zero_grad()
pred = model(x)
loss = loss_fn(pred, y)
loss.backward()
optimizer.step()

# print(model.weight)
weight_after = model.weight.detach().clone()
# print(weight_after)

print(f"weight before: {weight_before}")
print(f"weight after: {weight_after}")
print(f"changed: {not torch.allclose(weight_before, weight_after)}")
```

```python
def check_one_step_update(model, optimizer, loss_fn, x, y):
    first_param_before = next(model.parameters()).detach().clone()

    optimizer.zero_grad()
    pred = model(x)
    loss = loss_fn(pred, y)
    loss.backward()
    optimizer.step()

    first_param_after = next(model.parameters()).detach().clone()

    print(f"loss: {loss.item()}")
    print(f"parameter changed: {not torch.allclose(first_param_before, first_param_after)}")

check_one_step_update(model, optimizer, loss_fn, x, y)
```

```python
import torch
import torch.nn as nn
import torch.optim as optim

torch.manual_seed(42)

x = torch.randn(8, 3)
y = torch.randn(8, 1)

model = nn.Linear(3, 1)
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.05)

for step in range(3):
    #TODO 1. gradient 초기화
    optimizer.zero_grad()

    #TODO 2. forward
    pred = model(x)

    #TODO 3. loss 계산
    loss = loss_fn(pred, y)

    #TODO 4. backward
    loss.backward()
    for name, param in model.named_parameters():
        print(f"name: {name}\nparameters: {param}\n")

    #TODO 5. parameter update
    optimizer.step()

    #TODO 6. loss 출력
    print(f"step: {step}, loss: {loss.item()}\n")
```
