```python
import torch
import torch.nn as nn

class SimpleMLP(nn.Module):
    """
    28x28 흑백 이미지를 입력으로 받아 class 별 logits을 출력하는 간단한 MLP 모델
    """
    def __init__(self, input_dim=1*28*28, hidden_dim=128, num_classes=10):
        """
        모델에 필요한 layer를 정의
        
        input_dim: flatten 이후 샘플 하나가 가지는 feature 수
        hidden_dim: 은닉층의 출력 feature 수
        num_classes: 분류할 class 수
        """
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        """
        입력 Tensor x가 모델을 통과하는 순서를 정의
        """
        x = self.flatten(x)
        x = self.fc1(x)
        x = self.relu(x)
        logits = self.fc2(x)

        return logits
```

```python
images = torch.randn(8, 1, 28, 28)
model = SimpleMLP(input_dim=1*28*28, hidden_dim=128, num_classes=10)
logits = model(images)

print(f"images shape: {images.shape}")
print(f"logits shape: {logits.shape}")
```

```python
class DebugMLP(nn.Module):
    """
    forward 과정에서 중간 shape를 출력하는 디버깅용 MLP
    """
    def __init__(self, input_dim=784, hidden_dim=128, num_classes=10):
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, num_classes)

    def forward(self, x):
        print(f"input shape: {x.shape}")
        x = self.flatten(x)
        print(f"flatten shape: {x.shape}")
        x = self.fc1(x)
        print(f"fc1 shape: {x.shape}")
        x = self.relu(x)
        print(f"ReLU shape: {x.shape}")
        logits = self.fc2(x)
        print(f"fc2 shape: {logits.shape}")

        return logits

images = torch.randn(8, 1, 28, 28)
model = DebugMLP()
logits = model(images)
```

```python
def count_parameters(model):
    """
    학습 가능한 parameters 수를 계산
    """
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

def check_model_output(model, x, expected_shape):
    """
    모델 출력 shape가 기대한 shape와 같은지 확인
    """
    logits = model(x)

    print(f"input shape: {x.shape}")
    print(f"output shape: {logits.shape}")
    print(f"parameters: {count_parameters(model)}")
    assert logits.shape == expected_shape, f"출력 shape가 다릅니다."

    return logits

images = torch.randn(8, 1, 28, 28)
model = SimpleMLP(784, 64, 10)

logits = check_model_output(
    model = model,
    x = images,
    expected_shape = torch.Size([8, 10])
)
```

```python
# 오류 발생용 예시
images = torch.randn(8, 1, 28, 28)

model = SimpleMLP(28, 64, 10)

try:
    logits = model(images)
except RuntimeError as e:
    print(e)


flat = torch.flatten(images, start_dim=1)
print(f"images shape: {images.shape}")
print(f"flat shape: {flat.shape}")
print(f"flat feature: {flat.shape[-1]}")
print(f"model fc1 in_features: {model.fc1.in_features}")

model = SimpleMLP(input_dim=1*28*28, hidden_dim=64, num_classes=10)
logits = model(images)

print(f"\nlogits.shape: {logits.shape}")
```

```python
import torch
import torch.nn as nn

class ImageMLP(nn.Module):
    def __init__(self, input_dim=3*32*32, hidden_dim=100, num_classes=10):
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(in_features=input_dim, out_features=hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(in_features=hidden_dim, out_features=num_classes)

    def forward(self, x):
        x = self.flatten(x)
        x = self.fc1(x)
        x = self.relu(x)
        logits = self.fc2(x)

        return logits

images = torch.randn(12, 3, 32, 32)
model = ImageMLP()
logits = model(images)

print(f"logits shape: {logits.shape}")
```
