```python
import torch
import torch.nn as nn

class BatchNormMLP(nn.Module):
    def __init__(self, input_dim=784, hidden_dim=128, num_classes=10):
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.bn1 = nn.BatchNorm1d(hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, num_classes)
    def forward(self, x):
        x = self.flatten(x)
        x = self.fc1(x)
        x = self.bn1(x)
        x = self.relu(x)
        logits = self.fc2(x)
        return logits
```

4. 순서는 `Linear -> BatchNorm -> ReLU -> Linear`입니다.
5. 더미 입력을 넣었을 때 출력 shape는 `(8, 10)`이어야 합니다.

```python
class testMLP(nn.Module):
    def __init__(self, input_dim=784, hidden_dim=128, num_classes=10):
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.bn1 = nn.BatchNorm1d(hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, num_classes)
    def forward(self, x):
        x = self.flatten(x)
        x = self.fc1(x)
        x = self.bn1(x)
        x = self.relu(x)
        logits = self.fc2(x)
        return logits


dummy = torch.randn(8, 1, 28, 28)
model = testMLP()
logits = model(dummy)
print(f"logits shape: {logits.shape}")
```
