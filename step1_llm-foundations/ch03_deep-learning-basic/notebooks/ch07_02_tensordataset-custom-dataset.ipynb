```python
import torch
from torch.utils.data import Dataset

class TabularDataset(Dataset):
    def __init__(self, features, labels):
        self.features = features
        self.labels = labels
        assert len(features) == len(labels), "feature와 label의 샘플 수가 다릅니다."

    def __len__(self):
        return len(self.features)

    def __getitem__(self, idx):
        x = self.features[idx]
        y = self.labels[idx]
        return x, y
```

```python
X = torch.randn(12, 5)
y = torch.randint(low=0, high=3, size=(12,), dtype=torch.long)

dataset = TabularDataset(X, y)

print(f"dataset length: {len(dataset)}")

sample_x, sample_y = dataset[0]
print(f"sample_x shape: {sample_x.shape}")
print(f"sample_y: {sample_y}")
```

```python
from torch.utils.data import DataLoader

loader = DataLoader(
    dataset,
    batch_size = 4,
    shuffle = True,
)

batch_x, batch_y = next(iter(loader))
print(f"batch_x shape: {batch_x.shape}")
print(f"batch_y shape: {batch_y.shape}")
```

```python
import torch
from torch.utils.data import Dataset, DataLoader

# 다음 조건을 만족하는 Dataset을 작성하세요.
# 1. `features`와 `labels`를 입력으로 받습니다.
# 2. `__len__`은 전체 샘플 수를 반환합니다.
# 3. `__getitem__`은 `(x, y)`를 반환합니다.
# 4. DataLoader로 batch size 8의 batch를 만들고 shape를 출력합니다.

class MyclassificationDataset(Dataset):
    def __init__(self, features, labels):
        self.features = features
        self.labels = labels
    def __len__(self):
        return len(self.features)
    def __getitem__(self, idx):
        x = self.features[idx]
        y = self.labels[idx]
        return x, y

X = torch.randn(40, 6)
y = torch.randint(0, 3, size=(40,), dtype=torch.long)
dataset = MyclassificationDataset(X, y)

loader = DataLoader(
    dataset = dataset,
    batch_size = 8,
    shuffle = True,
)

batch_x, batch_y = next(iter(loader))
print(f"batch_x shape: {batch_x.shape}")
print(f"batch_y shape: {batch_y.shape}, batch_y dtype: {batch_y.dtype}")
```
