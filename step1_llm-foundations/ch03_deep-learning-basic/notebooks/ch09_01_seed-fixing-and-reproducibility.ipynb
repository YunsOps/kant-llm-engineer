```python
import random
import numpy as np
import torch

def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
```

```python
set_seed(42)
x1 = torch.randn(3)

set_seed(42)
x2 = torch.randn(3)

print(x1)
print(x2)
```

```python
from torch.utils.data import TensorDataset, DataLoader, random_split

set_seed(42)

X = torch.randn(100, 4)
y = torch.randint(0, 3, size=(100,))
dataset = TensorDataset(X, y)

split_generator = torch.Generator().manual_seed(42)
train_dataset, val_dataset = random_split(dataset, [80,20], generator=split_generator)

loader_generator = torch.Generator().manual_seed(42)
train_loader = DataLoader(train_dataset, batch_size=16, shuffle=True, generator=loader_generator)

first_batch_x, first_batch_y = next(iter(train_loader))
print(first_batch_x.shape, first_batch_y.shape)
```
